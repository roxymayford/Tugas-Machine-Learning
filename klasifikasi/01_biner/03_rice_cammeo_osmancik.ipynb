# Klasifikasi Biner: Rice (Cammeo and Osmancik)

Dataset Rice Cammeo and Osmancik berisi fitur morfologi butiran beras dari dua spesies bersertifikat asal Turki (Cammeo dan Osmancik) yang diperoleh melalui pemrosesan citra digital.

- Nama dataset: Rice (Cammeo and Osmancik)
- Sumber data: UCI Machine Learning Repository (ID: 545)
- Link dataset: https://archive.ics.uci.edu/dataset/545/rice+cammeo+and+osmancik
- Dimensi data: 3.810 baris dan 8 kolom (7 fitur morfologi dan 1 kolom target)
- Variabel target: Kolom `Class` yang berisi 2 kategori varietas beras: Cammeo dan Osmancik

```python
import pandas as pd
from ucimlrepo import fetch_ucirepo

rice = fetch_ucirepo(id=545)
X = rice.data.features
y = rice.data.targets
df = pd.concat([X, y], axis=1)

print("Shape data:", df.shape)
display(df.head())
print("\nTipe data:")
print(df.dtypes)
```

```python
import matplotlib.pyplot as plt

def plot_dist(y, title, rot=0):
    ax = pd.Series(y).value_counts().sort_index().plot(kind="bar", figsize=(6, 3), color="#4C78A8")
    ax.set_title(title)
    ax.set_ylabel("jumlah")
    plt.xticks(rotation=rot)
    plt.tight_layout()
    plt.show()

target_col = "Class"

print("Nilai unik target:", sorted(df[target_col].unique()), "=> biner:", df[target_col].nunique() == 2)
print("\nDistribusi kelas:")
print(df[target_col].value_counts())
plot_dist(df[target_col], "Rice: distribusi kelas")
```

```python
assert df[target_col].nunique() == 2, "Nilai unik target harus berjumlah 2"
assert set(df[target_col].unique()) == {"Cammeo", "Osmancik"}, "Nilai kategori target tidak sesuai"
assert df[target_col].isnull().sum() == 0, "Target mengandung nilai null"
assert df.shape == (3810, 8), f"Bentuk dataframe tidak sesuai: {df.shape}"

print("Semua verifikasi lolos.")
```

## Kesimpulan

Dataset Rice Cammeo and Osmancik terbukti merupakan klasifikasi biner. Nilai unik target `Class` berjumlah tepat 2, yaitu Osmancik (2.180 baris) dan Cammeo (1.630 baris), dengan tiap sampel beras tergolong ke dalam satu varietas.
